```python
# TASK 1.10
from pyspark.sql import SparkSession
import os

# Hadoop setup
os.environ["HADOOP_HOME"] = r"C:\hadoop"
os.environ["hadoop.home.dir"] = r"C:\Hadoop"

# MySQL Connector/J location
jar_path = r"C:\Users\barika.johnbazia\Desktop\PythonProjects\sba - module 345\mysql-connector-j-26.7.0.jar"

# Check that Python can find the JAR
print("JAR exists:", os.path.isfile(jar_path))

# Create Spark session with MySQL JDBC driver
spark = (
    SparkSession.builder
    .appName("CompanyABC Stock")
    .config("spark.jars", jar_path)
    .config("spark.driver.extraClassPath", jar_path)
    .getOrCreate()
)

print("Spark started successfully.")

try:
    spark._jvm.java.lang.Class.forName(
        "com.mysql.cj.jdbc.Driver"
    )

    print("SUCCESS - MySQL JDBC driver loaded!")

except Exception as e:
    print("FAILED - JDBC driver did not load")
    print(e)

company_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv("CompanyABC_stock.csv")
)

company_df.show()

company_df.write \
    .format("jdbc") \
    .option(
        "url",
        "jdbc:mysql://localhost:3306/CompanyABC_DB"
    ) \
    .option("dbtable", "companyabc_stock") \
    .option("user", "root") \
    .option("password", "welcome") \
    .option("driver", "com.mysql.cj.jdbc.Driver") \
    .mode("overwrite") \
    .save()

print("SUCCESS - Data written to CompanyABC_DB!")
```

```python
# Task 1.1
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
import mysql.connector
import os
 
# Hadoop setup for Windows
os.environ["HADOOP_HOME"] = r"C:\hadoop"
os.environ["hadoop.home.dir"] = r"C:\Hadoop"

spark = ( SparkSession.builder
    .appName("CompanyABC")
    .config("spark.jars.packages", "com.mysql-connector-j:26.7.0.jar")
    .getOrCreate()
)
CompanyABC_stock = r"C:\Users\barika.johnbazia\Desktop\PythonProjects\sba - module 345\CompanyABC_stock.csv"
spark_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(CompanyABC_stock)
)
spark_df.columns
spark_df.printSchema()
spark_df.describe().show()
spark_df.show()
```

```python
# TASK 1.2
spark_df.show(5)
```

```python
# TASK 1.3 

spark_df = spark_df.withColumn("HV Ratio", spark_df["High"] / spark_df["Volume"])
spark_df.select("HV Ratio").show(5)
 
```

```python
# TASK 1.4 Find out what day the stock price was the highest.
spark_df.orderBy(spark_df.High.desc()).show(1)
```

```python
# TASK 1.5 The avg of the closing price
from pyspark.sql.functions import avg
spark_df.select(avg("Close")).show()
```

```python
# TASK 1.6
from pyspark.sql.functions import max
from pyspark.sql.functions import min

spark_df.select(max("Volume")).show()
spark_df.select(min("Volume")).show()

```

```python
# TASK 1.7
task17 = spark_df.select(spark_df["Close"] < 70).count()
print("Closing Value Less than 70")
```

```python
# TASK 1.8
More80=int(spark_df.filter(spark_df["High"] > 80).count())
total_count = spark_df.count()
Answer = round((More80/total_count) * 100, 2)
print("Task 1.8:", Answer)
```
